### Invers Matriks (linalg.inv)


```python
persegi = np.array([[4, 7], [2, 6]])
inv = np.linalg.inv(persegi)

print("Matriks Asal:\n", persegi)
print("\nInvers Matriks:\n", inv)
print("\nVerifikasi (A @ A_inv = Identitas):\n", np.round(persegi @ inv))
```

### Determinan Matriks (linalg.det)

### Manajemen Memori (View vs Copy)


```python
induk = np.array([10, 20, 30, 40])
tampilan_view = induk[0:2]
salinan_copy = induk[0:2].copy()

tampilan_view[0] = 999
print("Array Induk (terpengaruh view):", induk)
print("Salinan Copy (tidak berubah) :", salinan_copy)
```

### Singular Value Decomposition (SVD)


```python
M_svd = np.array([[1, 2], [3, 4], [5, 6]])
U, S, Vt = np.linalg.svd(M_svd)

print("Matriks U :\n", np.round(U, 3))
print("Nilai Singular S:", np.round(S, 3))
print("Matriks V^T:\n", np.round(Vt, 3))
```

### QR Decomposition (linalg.qr)


```python
M_qr = np.array([[1, 2], [3, 4], [5, 6]])
Q, R = np.linalg.qr(M_qr)

print("Matriks Ortogonal Q:\n", np.round(Q, 3))
print("\nMatriks Segitiga Atas R:\n", np.round(R, 3))
```

### Cholesky Decomposition (linalg.cholesky)


```python
# Matriks simetris definit positif
A_pos = np.array([[4, 12, -16], [12, 37, -43], [-16, -43, 98]])
L = np.linalg.cholesky(A_pos)

print("Faktor Segitiga Bawah L:\n", L)
print("\nVerifikasi (L @ L.T = A):\n", L @ L.T)
```

### Vektor dan Matriks Norm (linalg.norm)


```python
vektor_v = np.array([3, 4])
matriks_m = np.array([[1, 2], [3, 4]])

print("L1 Norm (Manhattan) :", np.linalg.norm(vektor_v, ord=1))
print("L2 Norm (Euclidean) :", np.linalg.norm(vektor_v, ord=2))
print("Frobenius Norm      :", np.linalg.norm(matriks_m, 'fro'))
```

### Bilangan Kondisi Matriks (linalg.cond)


```python
A_stabil = np.array([[1, 2], [3, 4]])
A_buruk = np.array([[1, 2], [1, 2.0001]])

print("Condition number stabil:", round(np.linalg.cond(A_stabil), 2))
print("Condition number buruk :", round(np.linalg.cond(A_buruk), 2))
```

### Moore-Penrose Pseudoinverse (linalg.pinv)


```python
# Invers semu untuk matriks persegi-panjang
persegi_panjang = np.array([[1, 2], [3, 4], [5, 6]])
pinv = np.linalg.pinv(persegi_panjang)

print("Pseudoinverse:\n", np.round(pinv, 3))
print("\nVerifikasi (A @ pinv @ A = A):\n", np.round(persegi_panjang @ pinv @ persegi_panjang, 1))
```

### Polynomial Fitting (np.polyfit & poly1d)


```python
x_poly = np.array([0, 1, 2, 3, 4, 5])
y_poly = np.array([1.1, 2.9, 7.2, 13.8, 22.9, 34.0])

koefisien = np.polyfit(x_poly, y_poly, deg=2)
model_poli = np.poly1d(koefisien)

print("Persamaan Kuadratik Hasil Fit:\n", model_poli)
print(f"Prediksi untuk x=6: {model_poli(6):.2f}")
```

### Fast Fourier Transform 1D (np.fft.fft)


```python
t = np.linspace(0, 1, 100, endpoint=False)
sinyal = np.sin(2 * np.pi * 5 * t) + 0.5 * np.sin(2 * np.pi * 20 * t)

spektrum = np.fft.fft(sinyal)
frekuensi = np.fft.fftfreq(len(t), d=t[1]-t[0])

amplitudo = np.abs(spektrum)[:len(t)//2]
print(f"Frekuensi dominan terdeteksi: {frekuensi[:len(t)//2][np.argmax(amplitudo)]:.1f} Hz")
```

### Notasi Einstein Summation (np.einsum)


```python
# Perkalian matriks dengan einsum
M1 = np.array([[1, 2], [3, 4]])
M2 = np.array([[5, 6], [7, 8]])

einsum_matmul = np.einsum('ij,jk->ik', M1, M2)
print("Einsum Matmul:\n", einsum_matmul)
print("Einsum Trace :", np.einsum('ii', M1))
```

### Masked Array (numpy.ma)

### Grid Koordinat 2D (np.meshgrid)

### Vektorisasi Fungsi Logika (np.vectorize)


```python
def klasifikasi_usia(u):
    if u < 17:
        return 'Remaja'
    elif u < 60:
        return 'Dewasa'
    else:
        return 'Lansia'

vektor_klasifikasi = np.vectorize(klasifikasi_usia)
data_usia = np.array([12, 25, 45, 68, 16])
print(vektor_klasifikasi(data_usia))
```

### Gradien Numerik (np.gradient)


```python
# y = x^2, turunan analitik dy/dx = 2x
x_pts = np.array([0, 1, 2, 3, 4, 5])
y_pts = x_pts**2

grad = np.gradient(y_pts, x_pts)
print("Nilai y       :", y_pts)
print("Gradien dy/dx :", grad)
```

### Interpolasi Linier 1D (np.interp)


```python
x_diketahui = np.array([0, 2, 4, 6])
y_diketahui = np.array([0, 10, 20, 30])

x_baru = np.array([1, 3, 5])
y_interpolasi = np.interp(x_baru, x_diketahui, y_diketahui)

print("Titik prediksi x=1, 3, 5 -> y:", y_interpolasi)
```

### Operasi Bitwise pada Array
