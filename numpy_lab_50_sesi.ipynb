### Import NumPy & Cek Versi


```python
import numpy as np

print(f"NumPy version: {np.__version__}")
```

### Membuat Array 1D


```python
# Array satu dimensi dari list
arr_1d = np.array([10, 20, 30, 40, 50])
print(arr_1d)
```

### Membuat Array 2D & 3D


```python
# Matriks 2D dan tensor 3D
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
arr_3d = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

print("Array 2D:\n", arr_2d)
print("\nArray 3D:\n", arr_3d)
```

### Atribut Array (ndim, shape, size, dtype)


```python
# Inspeksi struktur array
print(f"Jumlah dimensi (ndim) : {arr_2d.ndim}")
print(f"Bentuk ukuran (shape) : {arr_2d.shape}")
print(f"Total elemen (size)   : {arr_2d.size}")
print(f"Tipe data (dtype)     : {arr_2d.dtype}")
```

### Array Nol & Satu (zeros & ones)


```python
zeros_arr = np.zeros((2, 3))
ones_arr = np.ones((2, 3), dtype=int)

print("Zeros:\n", zeros_arr)
print("\nOnes:\n", ones_arr)
```

### Array Kosong & Seragam (empty & full)


```python
empty_arr = np.empty((2, 2))
full_arr = np.full((2, 3), fill_value=7)

print("Full:\n", full_arr)
```

### Rentang Angka Berurutan (arange)


```python
# np.arange(start, stop, step)
rentang = np.arange(0, 20, 2)
print("arange(0, 20, 2):", rentang)
```

### Spasi Linear (linspace)


```python
# Pembagian rentang berjarak sama
lin = np.linspace(0, 1, 5)
print("linspace(0, 1, 5):", lin)
```

### Bilangan Acak Desimal (rand & randn)


```python
np.random.seed(42)
rand_uniform = np.random.rand(2, 3)
rand_normal = np.random.randn(2, 3)

print("Uniform (0-1):\n", rand_uniform)
print("\nNormal (Mean 0, Std 1):\n", rand_normal)
```

### Bilangan Acak Bulat (randint)


```python
rand_int = np.random.randint(1, 100, size=(3, 3))
print("Integer acak (1-99):\n", rand_int)
```

### Indeksasi Array 1D


```python
arr = np.array([10, 20, 30, 40, 50])
print("Elemen pertama :", arr[0])
print("Elemen terakhir:", arr[-1])
```

### Indeksasi Matriks 2D


```python
matriks = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])
print("Baris 1 Kolom 2:", matriks[0, 1])
print("Baris terakhir :", matriks[-1, -1])
```

### Slicing Array 1D


```python
data = np.arange(10, 100, 10)
print("Original   :", data)
print("Index 2-5  :", data[2:6])
print("Lompat 2   :", data[::2])
print("Balik arah :", data[::-1])
```

### Slicing Matriks 2D


```python
grid = np.arange(1, 17).reshape(4, 4)
print("Matriks 4x4:\n", grid)
print("\nSub-matriks 2x2 (tengah):\n", grid[1:3, 1:3])
```

### Boolean Indexing


```python
nilai = np.array([45, 78, 92, 55, 88, 62])
lulus = nilai[nilai >= 75]
print("Nilai >= 75:", lulus)
```

### Fancy Indexing


```python
vektor = np.array([100, 200, 300, 400, 500, 600])
indeks_pilihan = [0, 2, 4]
print("Elemen indeks [0, 2, 4]:", vektor[indeks_pilihan])
```

### Kondisional dengan np.where


```python
angka = np.array([12, -5, 20, -8, 0, 15])
hasil = np.where(angka > 0, "Positif", "Non-Positif")
print(hasil)
```

### Pencarian Posisi (argwhere & nonzero)


```python
skor = np.array([10, 50, 80, 20, 90])
posisi = np.argwhere(skor > 50)
print("Koordinat skor > 50:\n", posisi)
```

### Indeks Nilai Ekstrem (argmax & argmin)


```python
deret = np.array([15, 82, 33, 99, 45, 12])
print(f"Index Max ({deret.max()}): {np.argmax(deret)}")
print(f"Index Min ({deret.min()}): {np.argmin(deret)}")
```

### Pengurutan Indeks (argsort)


```python
data_acak = np.array([40, 10, 50, 20, 30])
urutan_indeks = np.argsort(data_acak)
print("Indeks terurut:", urutan_indeks)
print("Array terurut :", data_acak[urutan_indeks])
```

### Mengubah Bentuk Array (reshape)


```python
linear = np.arange(12)
matriks_3x4 = linear.reshape(3, 4)
print("Bentuk 3x4:\n", matriks_3x4)
```

### Meratakan Array (ravel & flatten)


```python
print("Ravel (view)   :", matriks_3x4.ravel())
print("Flatten (copy) :", matriks_3x4.flatten())
```

### Transpose Matriks (.T)


```python
m = np.array([[1, 2, 3], [4, 5, 6]])
print("Sebelum Transpose:\n", m)
print("\nSetelah Transpose (.T):\n", m.T)
```

### Menambah Dimensi (newaxis & expand_dims)


```python
v = np.array([1, 2, 3])
v_col = v[:, np.newaxis]
v_exp = np.expand_dims(v, axis=0)

print("Bentuk asal      :", v.shape)
print("Bentuk baris baru:", v_exp.shape)
print("Bentuk kolom baru:", v_col.shape)
```

### Menghapus Dimensi Berukuran Satu (squeeze)


```python
tensor_1 = np.ones((1, 3, 1))
squeezed = np.squeeze(tensor_1)
print("Bentuk asal   :", tensor_1.shape)
print("Setelah squeeze:", squeezed.shape)
```

### Penggabungan Array (concatenate)


```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

concat_v = np.concatenate([a, b], axis=0)
concat_h = np.concatenate([a, b], axis=1)

print("Concatenate Axis 0:\n", concat_v)
print("\nConcatenate Axis 1:\n", concat_h)
```

### Stacking Array (vstack & hstack)


```python
x = np.array([1, 2, 3])
y = np.array([4, 5, 6])

print("vstack:\n", np.vstack([x, y]))
print("\nhstack:\n", np.hstack([x, y]))
```

### Memecah Array (split)


```python
besar = np.arange(9)
pecah = np.split(besar, 3)
print("Array hasil split 3 bagian:")
for p in pecah:
    print(p)
```

### Menambah Elemen (append & insert)


```python
dasar = np.array([10, 20, 30])
tambah_akhir = np.append(dasar, [40, 50])
sisip_tengah = np.insert(dasar, 1, 15)

print("Setelah append:", tambah_akhir)
print("Setelah insert:", sisip_tengah)
```

### Menghapus Elemen (delete)


```python
arr_del = np.array([10, 20, 30, 40, 50])
bersih = np.delete(arr_del, [1, 3])
print("Setelah elemen indeks 1 dan 3 dihapus:", bersih)
```

### Operasi Aritmatika Dasar


```python
p = np.array([10, 20, 30])
q = np.array([2, 4, 5])

print("Penjumlahan:", p + q)
print("Pengurangan:", p - q)
print("Perkalian  :", p * q)
print("Pembagian  :", p / q)
```

### Broadcasting Array


```python
mat = np.array([[1, 2, 3], [4, 5, 6]])
skalar = 10
vekt = np.array([100, 200, 300])

print("Matriks + Skalar:\n", mat + skalar)
print("\nMatriks + Vektor (Broadcasting):\n", mat + vekt)
```

### Universal Functions (ufuncs)


```python
val = np.array([1, 4, 9, 16])
print("Akar kuadrat (sqrt):", np.sqrt(val))
print("Eksponensial (exp) :", np.exp(val[:2]))
```

### Fungsi Trigonometri


```python
sudut_rad = np.array([0, np.pi/4, np.pi/2])
print("Sin:", np.round(np.sin(sudut_rad), 4))
print("Cos:", np.round(np.cos(sudut_rad), 4))
print("Tan:", np.round(np.tan(sudut_rad[:2]), 4))
```

### Fungsi Logaritma


```python
num = np.array([1, 10, 100])
print("Log Natural (ln):", np.log(num))
print("Log Basis 10    :", np.log10(num))
print("Log Basis 2     :", np.log2(num))
```

### Agregasi Dasar (sum, mean, median)


```python
sampel = np.array([10, 25, 40, 55, 70])
print(f"Total (sum) : {np.sum(sampel)}")
print(f"Rata-rata   : {np.mean(sampel)}")
print(f"Median      : {np.median(sampel)}")
```

### Varians dan Standar Deviasi (var & std)


```python
data_stat = np.array([12, 18, 24, 30, 36])
print(f"Varians         : {np.var(data_stat):.2f}")
print(f"Standar Deviasi : {np.std(data_stat):.2f}")
```

### Operasi Berbasis Axis


```python
tabel = np.array([[10, 20, 30], [40, 50, 60]])
print("Sum Kolom (axis=0):", np.sum(tabel, axis=0))
print("Sum Baris (axis=1):", np.sum(tabel, axis=1))
```

### Pembulatan Nilai (round, ceil, floor)


```python
desimal = np.array([1.23, 2.78, 3.50, -1.82])
print("Round (1 desimal):", np.round(desimal, 1))
print("Ceil (ke atas)   :", np.ceil(desimal))
print("Floor (ke bawah) :", np.floor(desimal))
```

### Perkalian Matriks (dot & @)


```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

hasil_matmul = A @ B
print("Hasil Perkalian Matriks (A @ B):\n", hasil_matmul)
```

### Menemukan Nilai Unik (unique)


```python
koleksi = np.array([1, 2, 2, 3, 4, 4, 4, 5])
unik, hitung = np.unique(koleksi, return_counts=True)
print("Elemen unik :", unik)
print("Frekuensi   :", hitung)
```

### Operasi Himpunan (intersect1d & union1d)


```python
set1 = np.array([1, 2, 3, 4, 5])
set2 = np.array([3, 4, 5, 6, 7])

print("Irisan (intersect1d):", np.intersect1d(set1, set2))
print("Gabungan (union1d)  :", np.union1d(set1, set2))
```

### Penanganan Nilai NaN


```python
data_nan = np.array([10, np.nan, 30, 40, np.nan])
print("Deteksi NaN             :", np.isnan(data_nan))
print("Rata-rata aman (nanmean):", np.nanmean(data_nan))
print("Total aman (nansum)     :", np.nansum(data_nan))
```

### Invers Matriks (linalg.inv)


```python
persegi = np.array([[4, 7], [2, 6]])
inv = np.linalg.inv(persegi)

print("Matriks Asal:\n", persegi)
print("\nInvers Matriks:\n", inv)
print("\nVerifikasi (A @ A_inv = Identitas):\n", np.round(persegi @ inv))
```

### Determinan Matriks (linalg.det)


```python
mat_det = np.array([[3, 8], [4, 6]])
det = np.linalg.det(mat_det)
print("Determinan:", round(det, 4))
```

### Eigenvalues dan Eigenvectors (linalg.eig)


```python
sym_mat = np.array([[2, 1], [1, 2]])
eigenvalues, eigenvectors = np.linalg.eig(sym_mat)

print("Eigenvalues  :\n", eigenvalues)
print("\nEigenvectors :\n", eigenvectors)
```

### Sistem Persamaan Linear (linalg.solve)


```python
# 2x + y = 8
# x + 3y = 9
A_eq = np.array([[2, 1], [1, 3]])
b_eq = np.array([8, 9])

solusi = np.linalg.solve(A_eq, b_eq)
print("Solusi [x, y]:", solusi)
```

### Ekspor dan Impor File Teks (savetxt & loadtxt)


```python
data_csv = np.array([[1.5, 2.8, 3.1], [4.0, 5.2, 6.7]])
np.savetxt("sample_data.csv", data_csv, delimiter=",", fmt="%.2f")

loaded_csv = np.loadtxt("sample_data.csv", delimiter=",")
print("Data dimuat dari CSV:\n", loaded_csv)
```

### Format Biner NumPy (.npy & .npz)


```python
arr_save = np.arange(10, 60, 10)
np.save("array_data.npy", arr_save)

arr_loaded = np.load("array_data.npy")
print("Data dimuat dari file .npy:", arr_loaded)
```

### Manajemen Memori (View vs Copy)


```python
induk = np.array([10, 20, 30, 40])
tampilan_view = induk[0:2]
salinan_copy = induk[0:2].copy()

tampilan_view[0] = 999
print("Array Induk (terpengaruh view):", induk)
print("Salinan Copy (tidak berubah) :", salinan_copy)
```
